# **Connecting and WOrking with in memory database (SQLite)**

```python
import sqlite3
import pandas as pd
connection = sqlite3.connect(':memory')
cursor = connection.cursor() # create a cursor object  to execute sql command
```

```python
create_table_query = '''

CREATE TABLE table_name (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    email TEXT UNIQUE,
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP
);

'''
cursor.execute(create_table_query)
print("student is created ")
```

```python
insert_data_query = '''
INSERT INTO table_name (name, age, email)
VALUES
    ('Alice Smith', 25, 'alice@example.com'),
    ('Bob Jones', 30, 'bob@example.com'),
    ('Charlie Brown', 22, 'charlie@example.com'),
    ('David Miller', 28, 'david@example.com'),
    ('Emma Wilson', 24, 'emma@example.com'),
    ('Frank Thomas', 35, 'frank@example.com'),
    ('Grace Davis', 29, 'grace@example.com');
'''
cursor.execute(insert_data_query)
connection.commit()
print("data has been inserted..")
```

```python
query = 'SELECT * FROM table_name;'
df = pd.read_sql(query, connection)
df
```

```python
update = 'UPDATE table_name SET name = "Alice Johnson" WHERE id = 1;'
cursor.execute(update)
connection.commit()
print("data has been updated..")
```

```python
cursor.close()
connection.close()
```

# **Connecting MYSQl database using pymysql**

```python

# !pip install pymysql
import pymysql
host = '
port = 13043
user = ''
password = ''
database = 'defaultdb'
```

```python
connection = pymysql.connect(host=host,
                            port=port,
                            user=user,
                            password=password,
                            database=database,
                            cursorclass = pymysql.cursors.DictCursor # this will return result as Dictionary
                            )
cursor = connection.cursor()
print("connected to db ")
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS employees (
    id INT AUTO_INCREMENT PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    salary DECIMAL(10, 2) NOT NULL,
    department_id INT
);
"""

add_data_query = """
INSERT INTO employees (first_name, last_name, email, salary, department_id)
VALUES (%s, %s, %s, %s, %s)
ON DUPLICATE KEY UPDATE email=email;
"""

data_values = [
    ('Grace', 'Singh', 'grace.singh@example.com', 75000.00, 101),
    ('Alex', 'Rivera', 'alex.r@example.com', 68000.00, 102),
    ('Sam', 'Chen', 'sam.c@example.com', 82000.00, 101)
]

# 3. Execute
with connection.cursor() as cursor:
    # Step 1: Create Table
    cursor.execute(create_table_query)

    # Step 2: Add Data
    cursor.executemany(add_data_query, data_values)

    # Step 3: Commit
    connection.commit()

print("Table created and data added successfully!")
```

```python
with connection.cursor(pymysql.cursors.DictCursor) as cursor:
    cursor.execute("SELECT * FROM employees;")
    rows = cursor.fetchall()

    print("--- EMPLOYEES LIST ---")
    for row in rows:
        # print(f"ID: {row['id']} | Name: {row['first_name']} {row['last_name']} | Email: {row['email']} | Salary: ${row['salary']}")
        print(row)
```

```python
cursor.close()
connection.close()
```
